```python
# 3) اختيار الصوت وتجربة النطق: استمع، وغيّر VOICE أو PRON إن لزم
import edge_tts, asyncio
from IPython.display import Audio, display

async def say(text, path):
    await edge_tts.Communicate(text, VOICE, rate=RATE, pitch=PITCH).save(path)

# ب) نطق الرموز بالصوت المختار
tests = [LINES[6], LINES[8], LINES[17], LINES[27]]   # X ، e ، y ، x²
for i, t in enumerate(tests):
    p = f"test_{i}.mp3"
    await say(spoken(t), p)
    print(spoken(t)); display(Audio(p))
```

```python
# 4) صناعة الصوت كاملًا + ملف التوقيت، ثم التحميل
import json, os, zipfile, asyncio
import edge_tts
from IPython.display import Audio, display
from pydub import AudioSegment
from google.colab import files

async def line_audio(text, path):
    for attempt in range(5):
        try:
            try:
                com = edge_tts.Communicate(text, VOICE, rate=RATE, pitch=PITCH, boundary="WordBoundary")
            except TypeError:  # نسخ قديمة من edge-tts
                com = edge_tts.Communicate(text, VOICE, rate=RATE, pitch=PITCH)
            words, data = [], bytearray()
            async for ch in com.stream():
                if ch["type"] == "audio":
                    data += ch["data"]
                elif ch["type"] == "WordBoundary":
                    words.append({"text": ch["text"], "start": ch["offset"] / 1e7, "end": (ch["offset"] + ch["duration"]) / 1e7})
            open(path, "wb").write(data)
            return words
        except Exception as err:
            print("إعادة المحاولة:", err); await asyncio.sleep(2 + attempt * 2)
    raise RuntimeError("فشل توليد السطر: " + text)

os.makedirs("lines", exist_ok=True)
full = AudioSegment.silent(duration=int(LEAD * 1000))
out = []
for i, line in enumerate(LINES):
    path = f"lines/{i + 1:02d}.mp3"
    words = await line_audio(spoken(line), path)
    seg = AudioSegment.from_file(path)
    t0 = len(full) / 1000
    out.append({"i": i + 1, "start": t0, "end": t0 + len(seg) / 1000, "words": [{**w, "start": w["start"] + t0, "end": w["end"] + t0} for w in words]})
    pause = GAP + (PAUSE if "حَاوِلْ بِنَفْسِكَ" in line else 0)
    full += seg + AudioSegment.silent(duration=int(pause * 1000))
    print(f"{i + 1:02d}  {len(seg) / 1000:5.1f}s  {len(words)} كلمة")

full.export("voice.mp3", format="mp3", bitrate="192k")
json.dump({"voice": VOICE, "rate": RATE, "gap": GAP, "duration": len(full) / 1000, "lines": out}, open("timings.json", "w"), ensure_ascii=False, indent=1)
with zipfile.ZipFile("ex2_voice.zip", "w") as z:
    z.write("voice.mp3"); z.write("timings.json")
print("المدة:", round(len(full) / 1000), "ثانية")
display(Audio("voice.mp3"))
files.download("ex2_voice.zip")
```
