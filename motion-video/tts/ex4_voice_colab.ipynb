```python
# 3) محرك الصوت + بنك الحروف + تجربة النطق: استمع قبل الخلية 4
import re, io, asyncio, edge_tts
from IPython.display import Audio, display
from pydub import AudioSegment
from pydub.silence import detect_leading_silence

async def tts_piece(text, voice, rate):
    for attempt in range(5):
        try:
            try:
                com = edge_tts.Communicate(text, voice, rate=rate, pitch=PITCH, boundary="WordBoundary")
            except TypeError:  # نسخ قديمة من edge-tts
                com = edge_tts.Communicate(text, voice, rate=rate, pitch=PITCH)
            words, data = [], bytearray()
            async for ch in com.stream():
                if ch["type"] == "audio":
                    data += ch["data"]
                elif ch["type"] == "WordBoundary":
                    words.append({"text": ch["text"], "start": ch["offset"] / 1e7, "end": (ch["offset"] + ch["duration"]) / 1e7})
            return AudioSegment.from_file(io.BytesIO(bytes(data)), format="mp3"), words
        except Exception as err:
            print("إعادة المحاولة:", err); await asyncio.sleep(2 + attempt * 2)
    raise RuntimeError("فشل توليد: " + text)

def trim(seg, words, keep=40):
    # يقص الصمت في طرفي القطعة حتى تلتصق القطع بسلاسة
    a = max(0, detect_leading_silence(seg, silence_threshold=-45) - keep)
    b = max(0, detect_leading_silence(seg.reverse(), silence_threshold=-45) - keep)
    seg = seg[a:len(seg) - b]
    return seg, [{**w, "start": max(0, w["start"] - a / 1000), "end": max(0, w["end"] - a / 1000)} for w in words]

import numpy as np

def friction(seg, ms=180):
    """عدد إطارات الاحتكاك (طاقة فوق 2.5 kHz) في بداية القطعة: حرف v المنطوق فعلًا له احتكاك، والمبتلَع لا."""
    s = np.array(seg.set_channels(1).set_frame_rate(16000).get_array_of_samples(), dtype=float)[: 16 * ms]
    n = 0
    for i in range(0, len(s) - 400, 160):
        S = np.abs(np.fft.rfft(s[i:i + 400] * np.hanning(400))) ** 2
        f = np.fft.rfftfreq(400, 1 / 16000)
        n += S[(f > 2500) & (f < 7000)].sum() / (S.sum() + 1e-9) > 0.35
    return n

import os, base64
os.makedirs("letters", exist_ok=True)
os.makedirs("bank", exist_ok=True)
for k, b64 in LETTERS_B64.items():
    open(f"letters/{k}.mp3", "wb").write(base64.b64decode(b64))

async def bank_options(k, voice):
    """نسخ مرقّمة للحرف وحده، تُحفظ في bank/ حتى يكون المقطع المختار هو نفسه الذي سمعه الأستاذ."""
    tries = [("V.", RATE, False), ("V.", SLOW_RATE, False), ("V،", SLOW_RATE, False),
             ("ڤِي.", SLOW_RATE, False), ("أُو V، V.", RATE, True), ("أُو V، V.", SLOW_RATE, True)]
    out = []
    for i, (text, rate, last_word) in enumerate(tries):
        path = f"bank/{k}_{i}.mp3"
        if not os.path.exists(path):
            seg, ws = await tts_piece(text, voice, rate)
            if last_word and ws:  # آخر كلمة في العبارة: الحرف في آخر الكلام يُنطق كاملًا
                w = ws[-1]
                seg = seg[max(0, int(w["start"] * 1000) - 60): int(w["end"] * 1000) + 120]
            seg, _ = trim(seg, [])
            seg.export(path, format="mp3")
        out.append(path)
    return out

def letter_piece(k):
    path = f"letters/{k}.mp3"
    if not os.path.exists(path):
        raise SystemExit(f"⚠ لا يوجد {k}.mp3 في بنك الحروف: اختر رقمًا في V_BANK بالخلية 2 ثم Run all")
    return AudioSegment.from_file(path)

async def synth(text, slow_from=None):
    """سطر كامل بصوت VOICE. slow_from: من هذه العبارة إلى آخر السطر يُقرأ الكلام بسرعة SLOW_RATE.
    كل ⟦v⟧ (إن وُجد في FOREIGN) يُنطق بصوته المختار ويُلصق بنفس مستوى الصوت."""
    if slow_from is not None and slow_from in text:
        k = text.index(slow_from)
        chunks = [(text[:k], RATE), (text[k:], SLOW_RATE)]
    else:
        chunks = [(text, RATE)]
    chunks = [(t, r) for t, r in chunks if re.search(r"[\u0600-\u06FFA-Za-z0-9]", t)]
    if len(chunks) == 1 and "⟦" not in chunks[0][0]:
        return await tts_piece(chunks[0][0], VOICE, chunks[0][1])
    pieces = []  # (seg, words, foreign, pause_after_ms)
    for chunk, rate in chunks:
        for p in re.split(r"(⟦[A-Za-z]⟧)", chunk):
            if p.startswith("⟦"):
                seg = letter_piece(p[1])
                pieces.append([seg, [{"text": p[1], "start": 0.0, "end": len(seg) / 1000}], True, 70])
            elif re.search(r"[\u0600-\u06FFA-Za-z0-9]", p):
                seg, ws = await tts_piece(p.strip(" ،,"), VOICE, rate)
                pieces.append([*trim(seg, ws), False, 70])
        pieces[-1][3] = 300  # وقفة الفاصلة بين جزء السرعة العادية وجزء البطء
    ar = [s for s, _, f, _ in pieces if not f]
    level = sum(s.dBFS for s in ar) / len(ar) if ar else -20
    full, words = AudioSegment.empty(), []
    for seg, ws, foreign, pause in pieces:
        if foreign:
            seg = seg.apply_gain(level - seg.dBFS)
        t0 = len(full) / 1000
        words += [{**w, "start": w["start"] + t0, "end": w["end"] + t0} for w in ws]
        full += seg + AudioSegment.silent(duration=pause)
    return full, words

# أ) بنك الحروف: نسخ v المرقّمة (✓ = احتكاك واضح، أي أن v نُطق فعلًا)
if not os.path.exists("letters/v.mp3") or V_BANK is not None:
    opts = await bank_options("v", V_VOICE)
    for i, path in enumerate(opts):
        n = friction(AudioSegment.from_file(path))
        print(f"V_BANK = {i}   احتكاك: {n} {'✓' if n >= 3 else ''}"); display(Audio(path))
    if V_BANK is not None:
        AudioSegment.from_file(opts[V_BANK]).export("letters/v.mp3", format="mp3")
        print(f"✓ حُفظ v.mp3 من النسخة {V_BANK}. أرسله للمساعد ليُستعمل في كل الدروس القادمة.")
        try:
            from google.colab import files; files.download("letters/v.mp3")
        except Exception:
            pass
    else:
        raise SystemExit("اختر رقم أوضح نسخة وضعه في V_BANK بالخلية 2، ثم Run all")
else:
    print("بنك الحروف: v.mp3 موجود"); display(Audio("letters/v.mp3"))

# ب) جمل الحلقة كما ستُنطق: أول جملتين فيهما v ملصوقة (استمع جيدًا لـ v)
tests = [3, 27, 25, 4]
for i, n in enumerate(tests):
    t = LINES[n]
    seg, _ = await synth(spoken(t), SLOW.get(n))
    seg.export(f"test_{i}.mp3", format="mp3")
    print(spoken(t)); display(Audio(f"test_{i}.mp3"))
```
