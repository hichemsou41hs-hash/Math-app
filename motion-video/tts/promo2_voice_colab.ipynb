```
VOICE = "ar-SA-HamedNeural"   # بدائل: ar-EG-ShakirNeural ، ar-AE-HamdanNeural ، ar-JO-TaimNeural ، ar-DZ-IsmaelNeural
RATE = "+0%"     # سرعة إعلانية خفيفة
PITCH = "+0Hz"
GAP = 0.6         # صمت بين الجمل (ثوانٍ)
LEAD = 0.5
LINES = [
 "لكل أستاذ تعليم ثانوي، يدرّس في القسم أو في دروس الدعم، ولكل تلميذ يدرس في الثانوية، وخاصة تلاميذ البكالوريا.",
 "إليك مكتبة الأستاذ سوايسية هشام: أكثر من مئة وأربعين ملفًا، للسنة الأولى والثانية والثالثة ثانوي، مطابقة للمنهاج وللتوزيع السنوي الجديد.",
 "ملخصات، ومذكرات، وتمارين نموذجية مع الحلول، وتمارين البكالوريا مع الحلول، بثلاث صيغ: PDF ووورد ولاتاك، وكل الملفات قابلة للتعديل.",
 "في السنة الثالثة ثانوي، كل درس فيه أربعة ملفات، لجميع الشعب العلمية ولشعبة التسيير والاقتصاد، وهذه كل دروس السنة بالترتيب.",
 "نفتح الدالة الأسية: تمارين البكالوريا مع الحلول، والملف نفسه بثلاث صيغ: PDF ووورد ولاتاك.",
 "ثم الاحتمالات: تمارين نموذجية مع الحلول.",
 "ثم مذكرة المتتاليات العددية، وبعدها ملخص الأعداد المركبة.",
 "شعبة التسيير والاقتصاد: ثمانية دروس حسب التوزيع السنوي. نفتح ملخص المتتاليات العددية، ثم مذكرة الاشتقاقية والاستمرارية.",
 "ثم تمارين البكالوريا في الدوال الأصلية، وملخص النهايات ودراسة الدوال.",
 "ثم مذكرة الدالة اللوغاريتمية، وملخص الدالة الأسية، وتمارين الإحصاء.",
 "السنة الثانية ثانوي، الشعب العلمية: تسعة دروس، لكل درس مذكرة وملخص وتمارين بالحلول. نفتح ملخص الاشتقاقية.",
 "ثم مذكرة المرجح، وتمارين الاحتمالات، وملخص النهايات.",
 "السنة الأولى ثانوي، جذع مشترك علوم وتكنولوجيا: ستة دروس بنفس المحتوى، من الأعداد والحساب إلى الهندسة المستوية.",
 "لمعرفة كيفية اقتناء جميع الملفات، أرسل رسالة على فايسبوك أو إنستغرام، أو على واتساب عبر هذا الرقم."
]
# تصحيح النطق: الكلمة كما تظهر على الشاشة -> الكتابة التي تُنطق (تشكيل أو كتابة صوتية).
# كلمة واحدة مقابل كلمة واحدة حتى يبقى توقيت الكتابة صحيحًا.
FIX = {
 "أستاذ": "أُسْتَاذِ",
 "تعليم": "تَعْلِيمٍ",
 "يدرّس": "يُدَرِّسُ",
 "يدرس": "يَدْرُسُ",
 "إليك": "إِلَيْكَ",
 "الشعب": "الشُّعَب",
 "ولشعبة": "وَلِشُعْبَة",
 "شعبة": "شُعْبَة",
 "الأسية": "الأُسِّيَّة",
 "صيغ": "صِيَغٍ",
 "ملفًا": "مِلَفًّا",
 "ملفات": "مِلَفَّات",
 "الملفات": "المِلَفَّات",
 "والملف": "وَالمِلَفُّ",
 "ولاتاك": "وLaTeX",
 "ووورد": "وَوُورْدْ",
 "ملخصات": "مُلَخَّصَات",
 "ومذكرات": "وَمُذَكِّرَات",
 "مذكرة": "مُذَكِّرَة",
 "وملخص": "وَمُلَخَّص",
 "ملخص": "مُلَخَّص",
 "المرجح": "المُرَجِّح",
 "المركبة": "المُرَكَّبَة",
 "المتتاليات": "المُتَتَالِيَات",
 "اللوغاريتمية": "اللُّوغَارِيتْمِيَّة",
 "الاشتقاقية": "الاِشْتِقَاقِيَّة",
 "والاستمرارية": "وَالاِسْتِمْرَارِيَّة",
 "جذع": "جِذْع",
 "اقتناء": "اقْتِنَاء",
 "فايسبوك": "فَايْسْبُوكْ",
 "إنستغرام": "إِنْسْتَغْرَامْ",
 "واتساب": "وَاتْسَابْ",
 "ثانوي": "ثَانَوِي",
}
def spoken(line):
    return " ".join(FIX.get(w.strip("،:؟.!"), w.strip("،:؟.!")) + w[len(w.rstrip("،:؟.!")):] if w.strip("،:؟.!") in FIX else w for w in line.split())
```

```
import io, asyncio, edge_tts
from IPython.display import Audio, display
from pydub import AudioSegment
from pydub.silence import detect_leading_silence

async def tts(text, voice=None, rate=None):
    voice, rate = voice or VOICE, rate or RATE
    for attempt in range(5):
        try:
            try:
                com = edge_tts.Communicate(text, voice, rate=rate, pitch=PITCH, boundary="WordBoundary")
            except TypeError:
                com = edge_tts.Communicate(text, voice, rate=rate, pitch=PITCH)
            words, data = [], bytearray()
            async for ch in com.stream():
                if ch["type"] == "audio": data += ch["data"]
                elif ch["type"] == "WordBoundary":
                    words.append({"text": ch["text"], "start": ch["offset"] / 1e7, "end": (ch["offset"] + ch["duration"]) / 1e7})
            seg = AudioSegment.from_file(io.BytesIO(bytes(data)), format="mp3")
            a = max(0, detect_leading_silence(seg, silence_threshold=-45) - 40)
            b = max(0, detect_leading_silence(seg.reverse(), silence_threshold=-45) - 40)
            seg = seg[a:len(seg) - b]
            return seg, [{**w, "start": max(0, w["start"] - a / 1000), "end": max(0, w["end"] - a / 1000)} for w in words]
        except Exception as err:
            print("إعادة المحاولة:", err); await asyncio.sleep(2 + 2 * attempt)
    raise RuntimeError(text)

for v in ["ar-SA-HamedNeural", "ar-EG-ShakirNeural", "ar-AE-HamdanNeural", "ar-JO-TaimNeural", "ar-DZ-IsmaelNeural"]:
    seg, _ = await tts(LINES[0], v)
    seg.export(f"{v}.mp3", format="mp3"); print(v); display(Audio(f"{v}.mp3"))
```

```
import json, zipfile
from google.colab import files
full = AudioSegment.silent(duration=int(LEAD * 1000)); out = []
for i, line in enumerate(LINES):
    seg, words = await tts(spoken(line))
    t0 = len(full) / 1000
    out.append({"i": i + 1, "text": line, "start": t0, "end": t0 + len(seg) / 1000,
                "words": [{**w, "start": w["start"] + t0, "end": w["end"] + t0} for w in words]})
    full += seg + AudioSegment.silent(duration=int(GAP * 1000))
    print(f"{i + 1}  {len(seg) / 1000:5.1f}s")
full.export("voice.mp3", format="mp3", bitrate="192k")
json.dump({"voice": VOICE, "rate": RATE, "duration": len(full) / 1000, "lines": out}, open("timings.json", "w"), ensure_ascii=False, indent=1)
with zipfile.ZipFile("promo2_voice.zip", "w") as z:
    z.write("voice.mp3"); z.write("timings.json")
print("المدة:", round(len(full) / 1000), "ثانية"); display(Audio("voice.mp3"))
files.download("promo2_voice.zip")
```
