```
VOICE = "ar-SA-HamedNeural"   # بدائل: ar-EG-ShakirNeural ، ar-AE-HamdanNeural ، ar-JO-TaimNeural ، ar-DZ-IsmaelNeural
RATE = "+0%"     # سرعة إعلانية خفيفة
PITCH = "+0Hz"
GAP = 0.6         # صمت بين الجمل (ثوانٍ)
LEAD = 0.5
LINES = [
 "لكل أستاذ تعليم ثانوي يدرّس في القسم أو في دروس الدعم، ولكل تلميذ في الثانوي، خاصة تلاميذ البكالوريا.",
 "إليك مكتبة الأستاذ سوايسية هشام: أكثر من مئة وأربعين ملفًا، مطابقة للمنهاج وللتوزيع السنوي الجديد.",
 "ملخصات، ومذكرات، وسلاسل تمارين نموذجية، وتمارين البكالوريا، جميعها بالحل المفصل، وبثلاث صيغ: بي دي إف ووورد ولاتاك.",
 "في السنة الثالثة ثانوي، كل درس يحتوي على أربعة ملفات: المذكرة، وملخص الدرس، وسلسلة تمارين نموذجية مرفقة بالحل، وحلول تمارين البكالوريا لجميع الشعب العلمية، وشعبة التسيير والاقتصاد.",
 "نأخذ مثال الدالة الأسية: مذكرة للدرس، وملخص شامل، وسلسلة تمارين نموذجية محلولة، وتمارين البكالوريا بالحل المفصل، بصيغ بي دي إف ووورد ولاتاك.",
 "الدالة اللوغاريتمية: مذكرة للدرس، وملخص شامل، وسلسلة تمارين بالحلول، وأكثر من مئة وسبعين صفحة من تمارين البكالوريا المحلولة.",
 "الدوال الأصلية والحساب التكاملي: مذكرة للدرس، وملخص شامل، وسلسلة تمارين نموذجية محلولة، وتمارين البكالوريا بالحل المفصل.",
 "الاحتمالات: أشجار منسقة، وجداول قوانين الاحتمال، وتمارين البكالوريا بالحل المفصل.",
 "وبنفس الطريقة: النهايات، والمتتاليات، والأعداد المركبة، والأعداد والحساب، والهندسة في الفضاء.",
 "شعبة التسيير والاقتصاد: المتتاليات العددية، مذكرة وملخص للدرس، وتمارين البكالوريا بالحل المفصل.",
 "والدوال الأصلية والحساب التكاملي: مذكرة وملخص للدرس، وتمارين البكالوريا بالحل المفصل.",
 "وبنفس الطريقة: الدالة اللوغاريتمية، والإحصاء، والاحتمالات.",
 "ننتقل إلى السنة الثانية ثانوي، الشعب العلمية: المذكرات، والملخصات، وسلاسل التمارين بالحلول.",
 "الاشتقاقية، والمرجح، والاحتمالات، والنهايات: ملخصات وتمارين بالحلول المفصلة.",
 "ثم السنة الأولى ثانوي، جذع مشترك علوم وتكنولوجيا: المذكرات، والملخصات، وسلاسل التمارين بالحلول.",
 "للحصول على جميع الملفات، أرسل رسالة في الخاص على صفحة الأستاذ سوايسية هشام للرياضيات، أو على واتساب عبر هذا الرقم."
]
# تصحيح النطق: الكلمة كما تظهر على الشاشة -> الكتابة التي تُنطق (تشكيل أو كتابة صوتية).
# كلمة واحدة مقابل كلمة واحدة حتى يبقى توقيت الكتابة صحيحًا.
FIX = {
 "أستاذ": "أُسْتَاذْ",
 "تعليم": "تَعْلِيمْ",
 "إليك": "إِلَيْكَ",
 "الشعب": "الشُّعَبِ",
 "لشعب": "لِشُعَبِ",
 "الأسية": "الأُسِّيَّةُ",
 "بصيغ": "بِصِيَغِ",
 "صيغ": "صِيَغٍ",
 "لاتاك": "LaTeX",
 "ولاتاك": "وLaTeX",
 "وورد": "وُورْدْ",
 "ووورد": "وَوُورْدْ",
 "بملخص": "بِمُلَخَّصْ",
 "جذع": "جِذْعْ",
}
def spoken(line):
    return " ".join(FIX.get(w.strip("،:؟.!"), w.strip("،:؟.!")) + w[len(w.rstrip("،:؟.!")):] if w.strip("،:؟.!") in FIX else w for w in line.split())
```

```
import io, asyncio, edge_tts
from IPython.display import Audio, display
from pydub import AudioSegment
from pydub.silence import detect_leading_silence

async def tts(text, voice=None, rate=None):
    voice, rate = voice or VOICE, rate or RATE
    for attempt in range(5):
        try:
            try:
                com = edge_tts.Communicate(text, voice, rate=rate, pitch=PITCH, boundary="WordBoundary")
            except TypeError:
                com = edge_tts.Communicate(text, voice, rate=rate, pitch=PITCH)
            words, data = [], bytearray()
            async for ch in com.stream():
                if ch["type"] == "audio": data += ch["data"]
                elif ch["type"] == "WordBoundary":
                    words.append({"text": ch["text"], "start": ch["offset"] / 1e7, "end": (ch["offset"] + ch["duration"]) / 1e7})
            seg = AudioSegment.from_file(io.BytesIO(bytes(data)), format="mp3")
            a = max(0, detect_leading_silence(seg, silence_threshold=-45) - 40)
            b = max(0, detect_leading_silence(seg.reverse(), silence_threshold=-45) - 40)
            seg = seg[a:len(seg) - b]
            return seg, [{**w, "start": max(0, w["start"] - a / 1000), "end": max(0, w["end"] - a / 1000)} for w in words]
        except Exception as err:
            print("إعادة المحاولة:", err); await asyncio.sleep(2 + 2 * attempt)
    raise RuntimeError(text)

for v in ["ar-SA-HamedNeural", "ar-EG-ShakirNeural", "ar-AE-HamdanNeural", "ar-JO-TaimNeural", "ar-DZ-IsmaelNeural"]:
    seg, _ = await tts(LINES[0], v)
    seg.export(f"{v}.mp3", format="mp3"); print(v); display(Audio(f"{v}.mp3"))
```

```
import json, zipfile
from google.colab import files
full = AudioSegment.silent(duration=int(LEAD * 1000)); out = []
for i, line in enumerate(LINES):
    seg, words = await tts(spoken(line))
    t0 = len(full) / 1000
    out.append({"i": i + 1, "text": line, "start": t0, "end": t0 + len(seg) / 1000,
                "words": [{**w, "start": w["start"] + t0, "end": w["end"] + t0} for w in words]})
    full += seg + AudioSegment.silent(duration=int(GAP * 1000))
    print(f"{i + 1}  {len(seg) / 1000:5.1f}s")
full.export("voice.mp3", format="mp3", bitrate="192k")
json.dump({"voice": VOICE, "rate": RATE, "duration": len(full) / 1000, "lines": out}, open("timings.json", "w"), ensure_ascii=False, indent=1)
with zipfile.ZipFile("promo2_voice.zip", "w") as z:
    z.write("voice.mp3"); z.write("timings.json")
print("المدة:", round(len(full) / 1000), "ثانية"); display(Audio("voice.mp3"))
files.download("promo2_voice.zip")
```
