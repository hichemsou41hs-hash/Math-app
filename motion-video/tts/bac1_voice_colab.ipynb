```python
# 3) محرك الصوت + تجربة النطق: استمع قبل الخلية 4
import re, io, asyncio, edge_tts
from IPython.display import Audio, display
from pydub import AudioSegment
from pydub.silence import detect_leading_silence

async def tts_piece(text, voice, rate):
    for attempt in range(5):
        try:
            try:
                com = edge_tts.Communicate(text, voice, rate=rate, pitch=PITCH, boundary="WordBoundary")
            except TypeError:  # نسخ قديمة من edge-tts
                com = edge_tts.Communicate(text, voice, rate=rate, pitch=PITCH)
            words, data = [], bytearray()
            async for ch in com.stream():
                if ch["type"] == "audio":
                    data += ch["data"]
                elif ch["type"] == "WordBoundary":
                    words.append({"text": ch["text"], "start": ch["offset"] / 1e7, "end": (ch["offset"] + ch["duration"]) / 1e7})
            return AudioSegment.from_file(io.BytesIO(bytes(data)), format="mp3"), words
        except Exception as err:
            print("إعادة المحاولة:", err); await asyncio.sleep(2 + attempt * 2)
    raise RuntimeError("فشل توليد: " + text)

def trim(seg, words, keep=40):
    # يقص الصمت في طرفي القطعة حتى تلتصق القطع بسلاسة
    a = max(0, detect_leading_silence(seg, silence_threshold=-45) - keep)
    b = max(0, detect_leading_silence(seg.reverse(), silence_threshold=-45) - keep)
    seg = seg[a:len(seg) - b]
    return seg, [{**w, "start": max(0, w["start"] - a / 1000), "end": max(0, w["end"] - a / 1000)} for w in words]

import numpy as np

def _fric_frames(seg, ms):
    """أزمنة (بالميلي ثانية) إطارات الاحتكاك (طاقة فوق 2.5 kHz) في أول القطعة."""
    s = np.array(seg.set_channels(1).set_frame_rate(16000).get_array_of_samples(), dtype=float)[: 16 * ms]
    hits = []
    for i in range(0, len(s) - 400, 160):
        S = np.abs(np.fft.rfft(s[i:i + 400] * np.hanning(400))) ** 2
        f = np.fft.rfftfreq(400, 1 / 16000)
        if S[(f > 2500) & (f < 7000)].sum() / (S.sum() + 1e-9) > 0.35:
            hits.append(i // 16)
    return hits

def friction(seg, ms=180):
    """عدد إطارات الاحتكاك: حرف v المنطوق فعلًا له احتكاك، والمبتلَع لا."""
    return len(_fric_frames(seg, ms))

def heavy(seg, length=2.0, voicing=-8):
    """v مشدّدة ومجهورة: يطيل الاحتكاك (length مرة، وهو التشديد) ويخلط فيه طنينًا منخفضًا
    مأخوذًا من الحركة التالية (الجهر: هو ما يفرّق v عن f)."""
    hits = _fric_frames(seg, 300)
    if not hits or length is None:
        return seg
    a, b = hits[0], min(hits[-1] + 25, len(seg))
    fr = seg[a:b]
    xf = max(5, min(20, len(fr) // 4))
    long = fr
    while len(long) < len(fr) * length:
        long = long.append(fr, crossfade=xf)
    long = long[: int(len(fr) * length)]
    if voicing is not None and len(seg) - b > 60:
        base = seg[b:b + 120].low_pass_filter(300)
        buzz = base
        while len(buzz) < len(long):
            buzz = buzz.append(base, crossfade=min(30, len(base) // 3))
        buzz = buzz[: len(long)].fade_in(min(30, len(long) // 3))
        long = long.overlay(buzz.apply_gain(voicing + long.dBFS - buzz.dBFS))
    out = seg[:a] + long
    return out.append(seg[b:], crossfade=min(10, len(seg) - b, len(long)))

def fix_v(seg, words):
    """يشدّد كل V ويجهره في مكانه داخل الجملة، ويُزيح توقيت الكلمات التالية بمقدار الإطالة.
    يُرجع أيضًا الاحتكاك قبل المعالجة لكل V (✓ = نُطق فعلًا)."""
    words = [dict(w) for w in words]
    report = []
    for j in reversed(range(len(words))):
        if words[j]["text"].strip("،,.") not in ("V", "v"):
            continue
        s_ms, e_ms = int(words[j]["start"] * 1000), int(words[j]["end"] * 1000)
        region = seg[s_ms:e_ms]
        report.append(friction(region))
        new = heavy(region, V_HEAVY, V_VOICING)
        if V_BASS:
            new = new.overlay(new.low_pass_filter(400).apply_gain(V_BASS - 6))
        seg = seg[:s_ms].append(new, crossfade=min(5, s_ms, len(new))).append(seg[e_ms:], crossfade=min(5, len(seg) - e_ms, len(new)))
        delta = (len(new) - (e_ms - s_ms)) / 1000
        words[j]["end"] += delta
        for w in words[j + 1:]:
            w["start"] += delta; w["end"] += delta
    return seg, words, report[::-1]

async def synth(text, slow_from=None, show=False):
    """السطر يُولَّد كاملًا دفعة واحدة. slow_from: من هذه العبارة إلى آخر السطر بسرعة SLOW_RATE
    (القطع عند الفاصلة التي قبلها، وهي وقفة طبيعية). ثم يُعالَج كل V في مكانه."""
    if slow_from is not None and slow_from in text and text.index(slow_from) > 0:
        k = text.index(slow_from)
        chunks = [(text[:k].strip(" ،,"), RATE), (text[k:], SLOW_RATE)]
    elif slow_from is not None and slow_from in text:
        chunks = [(text, SLOW_RATE)]
    else:
        chunks = [(text, RATE)]
    full, words, report = AudioSegment.empty(), [], []
    for i, (chunk, rate) in enumerate(chunks):
        seg, ws = await tts_piece(chunk, VOICE, rate)
        if len(chunks) > 1:
            seg, ws = trim(seg, ws, keep=80)
        seg, ws, rep = fix_v(seg, ws)
        report += rep
        t0 = len(full) / 1000
        words += [{**w, "start": w["start"] + t0, "end": w["end"] + t0} for w in ws]
        full += seg
        if i + 1 < len(chunks):
            full += AudioSegment.silent(duration=280)  # وقفة الفاصلة
    if show and report:
        print("   v:", "  ".join(f"{n}{'✓' if n >= 3 else '⚠'}" for n in report))
    return full, words

# جمل الحلقة كما ستُنطق: تعريف f (إكسبونونسيال)، f بريم، y، g و ln
tests = [4, 18, 15, 52]
for i, n in enumerate(tests):
    t = LINES[n]
    seg, _ = await synth(spoken(t), SLOW.get(n), show=True)
    seg.export(f"test_{i}.mp3", format="mp3")
    print(spoken(t)); display(Audio(f"test_{i}.mp3"))
```
